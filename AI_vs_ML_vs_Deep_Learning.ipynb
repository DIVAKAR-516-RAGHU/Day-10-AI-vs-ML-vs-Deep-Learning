## 1. Artificial Intelligence (AI)

Artificial Intelligence (AI) is the broader field of computer science
that focuses on creating systems capable of performing tasks that
normally require human intelligence.

These tasks can include:

- Reasoning
- Problem solving
- Decision making
- Understanding language
- Perception
- Learning

### Examples of AI

- Virtual assistants
- Chess-playing systems
- Self-driving vehicle systems
- Expert systems
- AI-powered customer support

## 2. Machine Learning (ML)

Machine Learning is a subset of Artificial Intelligence.

ML allows computers to learn patterns from data and make predictions
or decisions without being explicitly programmed for every individual
case.

### Examples of Machine Learning

- Email spam detection
- Product recommendation systems
- Credit risk prediction
- House price prediction
- Fraud detection

## 3. Deep Learning (DL)

Deep Learning is a subset of Machine Learning that uses artificial
neural networks with multiple layers to learn complex patterns from
large amounts of data.

Deep Learning is particularly useful for tasks involving images,
audio, video, and natural language.

### Examples of Deep Learning

- Face recognition
- Image classification
- Speech recognition
- Object detection
- Autonomous driving systems

## 4. Relationship Between AI, ML and Deep Learning

Artificial Intelligence is the broadest concept.

Machine Learning is a subset of Artificial Intelligence.

Deep Learning is a subset of Machine Learning.

### Relationship

AI
└── Machine Learning
    └── Deep Learning

## 5. Comparison Table

| Feature | Artificial Intelligence | Machine Learning | Deep Learning |
|---|---|---|---|
| Meaning | Broad field of intelligent systems | Subset of AI that learns from data | Subset of ML using deep neural networks |
| Main goal | Perform intelligent tasks | Learn patterns and make predictions | Learn complex patterns automatically |
| Data requirement | Varies | Usually requires data | Usually requires large amounts of data |
| Human intervention | Can be rule-based or learning-based | Requires feature selection in many traditional approaches | Learns features automatically |
| Algorithms | Rule-based systems, ML, search, planning | Regression, decision trees, SVM, clustering | CNN, RNN, Transformers |
| Computing requirement | Varies | Moderate to high | Generally high |
| Example | Virtual assistant | Recommendation system | Face recognition |

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Training data
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Prediction
prediction = model.predict([[6]])

print("Predicted value for 6:", prediction[0])
```

## 8. Interview Questions

### 1. What is Artificial Intelligence?

Artificial Intelligence is the field of creating computer systems
that can perform tasks requiring human-like intelligence, such as
reasoning, learning, perception, and decision making.

### 2. What is Machine Learning?

Machine Learning is a subset of Artificial Intelligence in which
computers learn patterns from data to make predictions or decisions.

### 3. What is Deep Learning?

Deep Learning is a subset of Machine Learning that uses multi-layer
artificial neural networks to learn complex patterns.

### 4. What is the difference between ML and DL?

Traditional Machine Learning often requires human-designed features,
while Deep Learning can automatically learn useful features from raw
data using neural networks.

### 5. How are AI, ML and DL related?

AI is the broader field, ML is a subset of AI, and DL is a subset of ML.

## 9. Conclusion

Artificial Intelligence, Machine Learning, and Deep Learning are
closely related technologies.

AI represents the broader goal of creating intelligent systems.
Machine Learning provides methods for systems to learn from data,
while Deep Learning uses multi-layer neural networks to solve complex
learning problems.

Understanding their relationship helps in selecting the appropriate
technology for different real-world applications.
